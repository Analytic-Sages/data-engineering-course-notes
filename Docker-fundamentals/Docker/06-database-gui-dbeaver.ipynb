# Database Visualization with DBeaver

Now that you have successfully built a pipeline that ingests live Bitcoin transactions into a PostgreSQL database, running queries in a command line (`pgcli`) can feel limiting. 

In professional data engineering environments, we use graphical database clients (GUIs) to explore tables, inspect schemas, and write SQL queries. In this module, you will learn how to connect **DBeaver** to your PostgreSQL database running inside Docker.

## 1. What is DBeaver?

**DBeaver** is a free, open-source, multi-platform database tool for developers, database administrators, and data engineers. 

### Why DBeaver?
- **Universal Client:** It supports PostgreSQL, MySQL, SQLite, Oracle, Redshift, Snowflake, and almost any database you will encounter.
- **Visual Data Browser:** You can double-click tables to view, filter, sort, and edit records without writing a single line of SQL.
- **ER Diagrams:** It automatically generates Entity-Relationship diagrams to visualize database schemas.
- **SQL Editor:** Auto-completion, formatting, and query history make writing SQL comfortable.

## 2. Installing DBeaver

Installing DBeaver is simple and supports all major operating systems:

### Windows & macOS
1. Go to the [Official DBeaver Download Page](https://dbeaver.io/download/).
2. Download the installer for your operating system (Community Edition).
3. Run the installer and follow the on-screen setup prompts.


## 3. Connecting DBeaver to Postgres in Docker

Ensure that your PostgreSQL container is running before attempting to connect. You can check it in your terminal by running:
```bash
docker ps
```

Follow these simple steps to set up the connection:

### Step 1: Create a New Connection
1. Launch DBeaver.
2. Click the **Database** then **New Connection**.
3. Select **PostgreSQL** from the list of databases and click **Next**.

### Step 2: Connection Settings
Fill in the connection details corresponding to your Docker PostgreSQL environment:

| Setting | Value |
| :--- | :--- |
| **Host** | `localhost` (or your host machine IP) |
| **Port** | `5432` |
| **Database** | `data-set` |
| **Username** | `root` |
| **Password** | `root` |

### Step 3: Test and Finish
1. Click the **Test Connection** button in the bottom-left corner of the connection settings window.
2. If this is your first time using PostgreSQL in DBeaver, it will ask to download the PostgreSQL JDBC driver. Click **Download**.
3. Once the connection test succeeds, click **Finish**.

## 4. Visualizing Your Bitcoin Transactions

Once connected, navigate through the Database Navigator panel on the left:

1. Expand the `postgresql-localhost` connection.
2. Expand `Databases` -> `data-set` -> `Schemas` -> `public` -> `Tables`.


## 5. Running SQL Queries in DBeaver

You can also write custom SQL queries to analyze your dataset:

1. Press `Ctrl + ]` (or right-click `bitcoin_transactions` -> `SQL Editor` -> `Open SQL console`).
2. Paste the query below into the editor:

```sql
SELECT 
    DATE(timestamp) as transaction_date,
    COUNT(*) as total_transactions,
    ROUND(AVG(size), 2) as avg_size_bytes,
    ROUND(AVG(fee), 2) as avg_fee_satoshis
FROM bitcoin_transactions
GROUP BY transaction_date;
```
3. Press `Ctrl + Enter` (or click the orange play button on the left sidebar of the editor) to execute the query and view the results in the bottom grid.

### Try Other Queries!
Since the SQL Editor is a general-purpose console, you can write and execute any SQL query you'd like.
