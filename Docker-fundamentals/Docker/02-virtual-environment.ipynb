# Virtual Environment with uv and Data Pipeine

A virtual environment is an isolated Python environment that keeps a project's dependencies separate from the rest of your system. This allows each project to use its own package versions without affecting other projects.

Unlike pip, uv is a fast Python package and project manager that can create virtual environments, install dependencies, and run Python scripts without requiring you to manually activate the virtual environment.

# Install uv

If you don't already have uv installed, install it with:

```bash
pip install uv
```

Verify the installation:

```bash
uv --version
```

**Create a New Project**

Create a new project directory and navigate into it:

```bash
mkdir pipeline
cd pipeline
```

**initialize a virtual environment :**

```bash
uv init --python=3.13
```

**Installing Pandas:**

```bash
uv add pandas
```

Your project structure will look like this:

pipeline/

├── .venv/

├── pyproject.toml

├── uv.lock

└── main.py

uv creates a pyproject.toml file for managing dependencies and a .python-version file.

Comparing Python Versions
```bash 
uv run which python  # Python in the virtual environment
uv run python -V
```

```bash
which python        # System Python
python -V
```

# Mini Pipeline

Add this code to the main.py

```python
import pandas as pd

blocks = pd.DataFrame({
    "block_number": [101, 102, 103],
    "transactions": [145, 231, 198]
})

print(blocks)
print(f"Average transactions: {blocks['transactions'].mean()}")
```


**Running the script**

```bash
uv run main.py
```

**Important:** The pandas package is installed only inside this virtual environment, not in your global Python system. This means other projects on your machine will not be affected.
