# Running Postgres With Docker

Now we move into real data engineering by using a containerized PostgreSQL database. Instead of installing Postgres manually, we run it inside Docker, where it is configured using environment variables and storage volumes.

First, create a folder that will hold database data (for persistence), for example:

```bash
mkdir pipeline_postgres_data
```

**Then run PostgreSQL using Docker:**

```bash
docker run -it \
  -e POSTGRES_USER="root" \
  -e POSTGRES_PASSWORD="root" \
  -e POSTGRES_DB="data-set" \
  -v $(pwd)/pipeline_postgres_data:/var/lib/postgresql \
  -p 5432:5432 \
  postgres:18
```

**What each parameter means:**

-e → sets environment variables like username, password, and database name

-$(pwd)/pipeline_postgres_data:/var/lib/postgresql  → Bind mount the pipeline_postgres_data folder to the docker
Data stored in the volume remains even after the container is removed

-p 5432:5432 → maps PostgreSQL port from container to your local machine

postgres:18 → uses PostgreSQL version 18 image

**Connecting to PostgreSQL**

Once the container is running, we can connect using pgcli.

Installing pgcli

```bash
uv add pgcli

uv add psycopg_binary
```



**Run the Postgres client:**

```bash
uv run pgcli -h localhost -p 5432 -u root -d data-set
```

What the flags mean:

-h → host (localhost since it's running locally)

-p → port

-u → username

-d → database name

You will be prompted for a password—enter: root


**Basic SQL Commands**

Once connected, you can run SQL commands:

-- List tables
```bash
\dt
```
-- Create a table
```bash
CREATE TABLE test (id INTEGER, name VARCHAR(50));
```
-- Insert data
```bash
INSERT INTO test VALUES (1, 'Hello Docker');
```
-- Query data
```bash
SELECT * FROM test;
```

-- Exit
```bash
\q
```
